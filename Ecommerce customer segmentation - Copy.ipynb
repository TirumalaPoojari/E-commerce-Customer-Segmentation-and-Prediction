# K-Means

```python
#K-Means
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

```python
#Elbow curve to get the right number of clusters
w = []
for i in range(1, 6):
    k = KMeans(n_clusters=i)
    k.fit(New_df_scaled)
    w.append(k.inertia_)
```

```python
plt.plot(range(1,6),w)
```

this Elbow Method used for visualize *and* which helps identify the optimal number of clusters for KMeans.

Inertia is a measure of how well the data points fit into their respective clusters

```python
k=KMeans(n_clusters=3,max_iter=50)
```

```python
k.fit(New_df_scaled)
```

```python
k.labels_
```

```python
k.predict([[5.1,3.5,1.4]])
```

```python
k.cluster_centers_
```

```python
print("Inertia:", k.inertia_)
```

```python
score = silhouette_score(New_df_scaled, k.labels_)
print("Silhouette Score:", score)
```

```python
plt.scatter(New_df_scaled.iloc[:, 0], New_df_scaled.iloc[:, 1], c=k.labels_, cmap='viridis')
plt.scatter(k.cluster_centers_[:, 0], k.cluster_centers_[:, 1], s=100, c='red', marker='X', label='Centroids')
plt.title('KMeans Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

k.cluster_centers_ This gives the coordinates of the centroids (the central points)

k.cluster_centers_[:, 0] This selects the x-coordinates of the centroids (first feature).

k.cluster_centers_[:, 1] This selects the y-coordinates of the centroids (second feature).

s=200: This adjusts the size of the markers (centroids) to 200 (larger than the default size).

c=red This colors the centroid markers in red.

marker=X This uses an X shape for the centroid markers.

label='Centroids' This provides a label for the centroids in the legend of the plot.

# Hierarchical Clustering

```python
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.datasets import make_blobs
from scipy.spatial.distance import pdist
```

```python
X = make_blobs(n_samples=10, centers=3, cluster_std=1.0, random_state=42)
```

```python
print(New_df_scaled.dtypes)
```

```python
New_df_scaled = New_df_scaled.select_dtypes(include=['number'])
```

```python
distance_matrix = pdist(New_df_scaled, metric='euclidean')
```

```python
Z = linkage(distance_matrix, method='ward')
```

```python
plt.figure(figsize=(10, 7))
dendrogram(Z)
plt.title("Dendrogram of Hierarchical Clustering")
plt.xlabel("Index")
plt.ylabel("Distance")
plt.show()
```

Small clusters merge into bigger clusters as you move up the diagram.

The higher the vertical line, the farther apart those clusters are

```python
from scipy.cluster.hierarchy import fcluster
clusters = fcluster(Z, t=3, criterion='maxclust')
print("Cluster assignments:", clusters)
```

```python
from sklearn.metrics import silhouette_score

# Calculate the silhouette score
sil_score = silhouette_score(New_df_scaled, clusters)
print("Silhouette Score:", sil_score)
```

```python
from sklearn.metrics import pairwise_distances_argmin_min
import numpy as np

# Calculate Inertia (Sum of squared distances of samples to their closest cluster center)
inertia = np.sum([np.sum((New_df_scaled[clusters == i] - New_df_scaled[clusters == i].mean(axis=0))**2) for i in np.unique(clusters)])
print("Inertia:", inertia)
```

```python
from scipy.cluster.hierarchy import cut_tree
cluster_labels=cut_tree(Z,n_clusters=3).flatten().flatten()
```

```python
pd.set_option('Display.Max_rows',None)
New_df_scaled['cluster_labels']=cluster_labels
New_df_scaled.head()
```

```python
sns.boxplot(x="cluster_labels", y="Recency", data=New_df_scaled)
plt.xlabel("Customer Segment (Cluster)")
plt.ylabel("Recency (Days Since Last Purchase)")
plt.title("Distribution of Recency Across Customer Segments")
plt.show()
```

**Cluster 0 (Left Box)**

Customers in this segment have a higher median recency (they purchased a long time ago).

This group has the largest spread, meaning customers vary in their purchasing habits.


**Cluster 1 (Middle Box)**

Customers in this segment tend to purchase more recently than Cluster 0.

There are many outliers, indicating that some customers behave differently.


 **Cluster 2 (Right Box)**

Customers in this segment have very low recency (they purchase very frequently).

The box is very small, meaning most customers in this segment behave similarly.
